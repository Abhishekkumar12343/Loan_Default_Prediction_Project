```python
from pyspark.sql import functions as F

df_silver = df_bronze

# --------------------------------------------------
# 1. Normalize empty strings
# --------------------------------------------------
for c, t in df_silver.dtypes:
    if t == "string":
        df_silver = df_silver.withColumn(
            c,
            F.when(F.trim(F.col(c)) == "", None).otherwise(F.col(c))
        )

# --------------------------------------------------
# 2. Derived fields
# --------------------------------------------------

# term_num
df_silver = df_silver.withColumn(
    "term_num",
    F.when(F.col("term").rlike("36"), 36)
     .when(F.col("term").rlike("60"), 60)
     .otherwise(None)
)

# emp length
df_silver = df_silver.withColumn(
    "emp_length_num",
    F.expr("""
        CASE
            WHEN emp_length IS NULL OR trim(emp_length) = '' THEN 0
            WHEN emp_length LIKE '%10+%' THEN 10
            WHEN emp_length LIKE '%< 1%' THEN 0
            WHEN emp_length RLIKE '^[0-9]+ year[s]?$' THEN try_cast(regexp_extract(emp_length, '([0-9]+)', 1) as int)
            ELSE try_cast(regexp_extract(emp_length, '([0-9]+)', 1) as int)
        END
    """)
)


# emp_length_num_filled (move after emp_length_num creation)
df_silver = df_silver.withColumn(
    "emp_length_num_filled",
    F.when(F.col("emp_length_num").isNull(), 0).otherwise(F.col("emp_length_num"))
)

# fico avg
df_silver = df_silver.withColumn(
    "fico_avg",
    (F.expr("try_cast(fico_range_low as double)") +
     F.expr("try_cast(fico_range_high as double)")) / 2
)

# installment to income
df_silver = df_silver.withColumn(
    "installment_to_income",
    F.when(
        F.expr("try_cast(annual_inc as double)") > 0,
        F.expr("try_cast(installment as double)") * 12 /
        F.expr("try_cast(annual_inc as double)")
    )
)

# --------------------------------------------------
# 3. Dates
# --------------------------------------------------

df_silver = (
    df_silver
    .withColumn("issue_dt", F.expr("try_to_date(issue_d, 'MMM-yyyy')"))
    .withColumn("earliest_cr_dt", F.expr("try_to_date(earliest_cr_line, 'MMM-yyyy')"))
    .withColumn("issue_year", F.year("issue_dt"))
    .withColumn("issue_month", F.month("issue_dt"))
    .withColumn(
        "credit_history_years",
        F.when(
            F.col("issue_dt").isNotNull() &
            F.col("earliest_cr_dt").isNotNull(),
            F.year("issue_dt") - F.year("earliest_cr_dt")
        )
    )
)

# --------------------------------------------------
# 4. Target
# --------------------------------------------------

bad_status = [
    "Charged Off",
    "Default",
    "Late (31-120 days)",
    "Late (16-30 days)"
]

df_silver = df_silver.withColumn(
    "target_default",
    F.when(F.col("loan_status").isin(bad_status), 1).otherwise(0)
)

# --------------------------------------------------
# 5. Base numeric casts (FINAL)
# --------------------------------------------------

double_cols = [
    "loan_amnt","funded_amnt","int_rate","installment",
    "annual_inc","dti","revol_bal","revol_util",
    "installment_to_income","fico_avg","pct_tl_nvr_dlq"
]

for c in double_cols:
    if c in df_silver.columns:
        df_silver = df_silver.withColumn(c, F.expr(f"try_cast({c} as double)"))

int_cols = [
    "term_num","emp_length_num","issue_year","issue_month",
    "credit_history_years","delinq_2yrs","inq_last_6mths",
    "open_acc","total_acc","acc_open_past_24mths","mort_acc",
    "num_rev_accts","num_actv_rev_tl","num_tl_30dpd",
    "num_tl_90g_dpd_24m","pub_rec_bankruptcies",
    "tax_liens","target_default"
]

for c in int_cols:
    if c in df_silver.columns:
        df_silver = df_silver.withColumn(c, F.expr(f"try_cast({c} as int)"))
```
